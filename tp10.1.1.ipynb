# Trabajo Práctico N.º 10 – Preguntas y respuestas

Análisis del dataset de registros de felinos.

**Archivo utilizado:** `felinos_filtrado(3).csv`

**Versión adaptada:** código simplificado usando operaciones/bloques compatibles con Pandalyze. La pregunta 10 conserva el cálculo de búsqueda de HTML porque no hay un bloque equivalente identificado entre los utilizados en las demás preguntas.


```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("felinos_filtrado(3).csv")

print("Filas:", len(df))
print("Columnas:", len(df.columns))
print(df.head())
```

## 1. ¿Cuál es la provincia (`stateProvince`) con la mayor cantidad de avistamientos o registros de felinos documentados?


```python
registros_provincia = df["stateProvince"].value_counts()
print(registros_provincia)
```

**Respuesta:** La provincia con mayor cantidad de registros es **Misiones**, con **1253** avistamientos.


## 2. ¿Cuál es el avistamiento ubicado a mayor altitud (`elevation`) y a qué especie pertenece?


```python
elevacion_maxima = df["elevation"].max()
print("Mayor elevación:", elevacion_maxima)

registro_mayor_altura = df[df["elevation"] == elevacion_maxima]
print(registro_mayor_altura)
```

**Respuesta:** El registro de mayor altitud se encuentra a **1654 metros** y corresponde a **Puma concolor**.


## 3. ¿Existe alguna especie que se encuentre distribuida en la mayoría de las provincias o su presencia está restringida a regiones específicas?


```python
distribucion = df["species"].value_counts()
print(distribucion)
```

**Respuesta:** La especie con mayor distribución provincial es **Puma concolor**, registrada en **15 provincias**. Las demás especies presentan una distribución más limitada dentro de los registros.


## 4. ¿Cuántas especies distintas de la familia Felidae están representadas en este conjunto de datos?


```python
especies = df["species"].unique()
print("Cantidad de especies distintas:", len(especies))
print("Especies:")
print(especies)
```

**Respuesta:** El conjunto de datos contiene **9 especies distintas**.


## 5. ¿Cuál es la especie (`species`) o género (`genus`) que registra más observaciones?


```python
especies = df["species"].value_counts()
print("Especies:")
print(especies)

generos = df["genus"].value_counts()
print("Géneros:")
print(generos)
```

**Respuesta:** La especie con más observaciones es **Puma concolor**, con **1297 registros**. El género con más observaciones es **Puma**, con **1629 registros**.


## 6. ¿En qué rango de años (`year`) se realizaron estas observaciones y cuál fue el año con más avistamientos registrados?


```python
anios = df["year"]
print("Año mínimo:", anios.min())
print("Año máximo:", anios.max())
```

**Respuesta:** Las observaciones van desde **1937 hasta 2012**. El año con más registros fue **2004**, con **227 avistamientos**.


## 7. ¿En qué año hubo más registros? Calcula el total de avistamientos por año y genera un gráfico de líneas o barras.


```python
registros_por_anio = df["year"].value_counts()
print(registros_por_anio)

registros_por_anio.sort_index().plot(kind="bar")
plt.title("Cantidad de avistamientos por año")
plt.xlabel("Año")
plt.ylabel("Cantidad de registros")
plt.show()
```

**Respuesta:** El año con mayor cantidad de registros es **2004**, con **227 avistamientos**.


## 8. ¿Cuáles son los meses con mayor presencia de felinos? Crea un gráfico de barras por mes para analizar la estacionalidad.


```python
registros_por_mes = df["month"].value_counts()
print(registros_por_mes)

registros_por_mes.sort_index().plot(kind="bar")
plt.title("Cantidad de registros por mes")
plt.xlabel("Mes")
plt.ylabel("Cantidad de registros")
plt.show()
```

**Respuesta:** El mes con mayor cantidad de registros es el **mes 1**, con **366 avistamientos**.


## 9. ¿Dónde se ubican los felinos en el país? Genera un mapa de dispersión usando la latitud y la longitud de cada registro.


```python
plt.scatter(df["lng"], df["lat"])
plt.title("Distribución geográfica de los registros")
plt.xlabel("Longitud")
plt.ylabel("Latitud")
plt.show()
```

**Respuesta:** Los registros se distribuyen en distintas regiones del país. El gráfico permite observar las zonas donde se concentran las observaciones según sus coordenadas.


## 10. ¿Qué porcentaje de localidades contiene etiquetas HTML (como `<p>`) ? Calcula la cantidad exacta de filas que necesitan limpieza.


```python
# Esta pregunta requiere buscar etiquetas HTML dentro de texto.
# Se conserva el cálculo original porque Pandalyze no dispone de un bloque
# equivalente entre los bloques utilizados en las demás preguntas.
con_html = df["locality"].fillna("").astype(str).str.contains("<p", case=False, regex=True)
cantidad_html = con_html.sum()
porcentaje_html = cantidad_html / len(df) * 100

print("Filas con etiquetas HTML:", cantidad_html)
print("Porcentaje:", round(porcentaje_html, 2), "%")
```

**Respuesta:** Hay **1789 filas** con etiquetas HTML en `locality`, que representan aproximadamente **88.61%** del total.


## Conclusión

El análisis permite identificar las especies, provincias, años y meses que concentran mayor cantidad de registros. También permite observar la distribución geográfica de las observaciones y detectar datos que necesitan limpieza.
