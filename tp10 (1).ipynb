```python
import pandas as pd
import plotly.express as px

felinos = pd.read_csv("felinos_filtrado.csv")
print(felinos.shape)
felinos.head()
```

## 1 Cual es la provincia con la mayor cantidad de avistamientos o registros de felinos documentados?

```python
felinos["stateProvince"].value_counts()
```

rt: la provincia con mas registros es Misiones, con 1253 avistamientos

## 2)¿Cuál es el avistamiento ubicado a mayor altitud (elevation) y a qué especie pertenece?

```python
elevacion_maxima = felinos["elevation"].max()
print(elevacion_maxima)

felinos[felinos["elevation"] == elevacion_maxima]
```

rt: el registro mas alto esta a 1654 metros y es de Puma concolor

## 3) ¿Existe alguna especie que se encuentre distribuida en la mayoría de las provincias o su presencia está restringida a regiones específicas?

```python
distribucion = felinos.groupby("species")["stateProvince"].nunique().sort_values(ascending=False)
distribucion
```

rt: Puma concolor es la que mas se distribuye, aparece en 15 provincias distintas. las demas especies estan mas restringidas a ciertas zonas

## 4) ¿Cuántas especies distintas de la familia Felidae están representadas en este conjunto de datos?

## 5) ¿Cuál es la especie (species) o género (genus) que registra más observaciones?

## 6) ¿En qué rango de años (year) se realizaron estas observaciones y cuál fue el año con más avistamientos registrados?

```python
print("min:", felinos["year"].min())
print("max:", felinos["year"].max())
```

rt: van desde 1937 hasta 2012

## 7) ¿En qué año hubo más registros? Calculá el total de avistamientos por año y generá un gráfico de líneas o barras.

```python
felinos_año = felinos["year"].value_counts().sort_index()
felinos_año
```

```python
px.bar(data_frame=felinos_año, x=felinos_año.index, y=felinos_año.values, title="Cantidad de avistamientos por año").show()
```

rt: el año con mas avistamientos fue 2004, con 227

## 8) ¿Cuáles son los meses con mayor presencia de felinos? Creá un gráfico de barras por mes para analizar la estacionalidad.

```python
felinos_mes = felinos["month"].value_counts().sort_index()
felinos_mes
```

```python
px.bar(data_frame=felinos_mes, x=felinos_mes.index, y=felinos_mes.values, title="Cantidad de registros por mes").show()
```

rt: enero (mes 1) es el que mas registros tiene, 366 en total

## 9) ¿Dónde se ubican los felinos en el país? Generá un mapa de dispersión usando la latitud y la longitud de cada registro.

```python
px.scatter(data_frame=felinos, x="lng", y="lat", color="species", title="Distribucion geografica de los registros").show()
```

rt: se puede ver como los registros se concentran mas que nada en el noreste (Misiones) y despues hay otro grupo importante en el sur, en Santa Cruz

## 10) ¿Qué porcentaje de localidades contiene etiquetas HTML (como <p>)? Calculá la cantidad exacta de filas que necesitan limpieza.

```python
con_html = felinos["locality"].fillna("").astype(str).str.contains("<p", case=False, regex=True)
cantidad_html = con_html.sum()
porcentaje_html = cantidad_html / len(felinos) * 100

print(cantidad_html)
print(round(porcentaje_html, 2))
```

rt: hay 1789 filas que tienen etiquetas html en locality, mas o menos el 88.61% del total. estas habria que limpiarlas antes de usar el campo para otra cosa

## Conclusion

La mayoria de los registros son de los años 2000, con un pico en 2004. Por ultimo, buena parte del campo locality tiene tags de html que habria que limpiar.
